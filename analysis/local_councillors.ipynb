```python
councillor_count_dict = {'year': np.arange(1999, 2026)}
for council, council_dict in councillors.items():
    years = np.concatenate(list(council_dict.values()))
    counts, lower_bin_bounds = np.histogram(years, bins=np.arange(1998.5, 2026.5))
    councillor_count_dict[council] = counts
```

```python
df = pd.DataFrame(councillor_count_dict)
df = df.set_index('year')
order = [
    'Huon Valley',
    'Clarence',
    'Hobart',
    'Kingborough',
    'Sorell',
    'Brighton',
    'Southern Midlands',
    'West Coast',
    'Launceston',
    'West Tamar',
    """Break O'Day""",
    'Devonport',
    'Glenorchy',
    'Tasman',
]
df = df[order]
df
```

```python
ax = df.plot(kind='bar', stacked=True, figsize=(10,5), width=1, colormap='tab20')
ax.legend(loc='upper left', bbox_to_anchor=(1.0, 1.0))
plt.title("Tasmanian Greens endorsed local councillors")
plt.ylabel("Number of councillors at year end")
plt.axvline(x=15, color='grey', linestyle=':')
plt.tight_layout()
plt.savefig('local_councillors.png', dpi=300, bbox_inches='tight')
plt.show()
```
