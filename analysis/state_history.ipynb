```python
#mp_count_dict = {'year': np.arange(1972, 2023)}
mp_count_dict = {'year': np.arange(1983, 2027)}
for state, mp_dict in state_mps.items():
    years = np.concatenate(list(mp_dict.values()))
    counts, lower_bin_bounds = np.histogram(years, bins=np.arange(1982.5, 2027.5))
    mp_count_dict[state] = counts
```

```python
df = pd.DataFrame(mp_count_dict)
df = df.set_index('year')
order = [
    'TAS Legislative Assembly',
    'TAS Legislative Council',
    'WA Legislative Assembly',
    'WA Legislative Council',
    'NSW Legislative Assembly',
    'NSW Legislative Council',
    'ACT Legislative Assembly',
    'SA House of Assembly',
    'SA Legislative Council',
    'VIC Legislative Assembly',
    'VIC Legislative Council',
    'QLD Legislative Assembly',
    'NT Legislative Assembly'
]
df = df[order]
df
```

The high water mark was 33 state/territory MPs at the end of August 2024 (between then and the end of 2024 the Greens lost 1 seat at the QLD state election, 2 seats at the ACT election and Sam Hibbins stepped down in Victoria).

## Victorian summary

```python
vic_df = df[['VIC Legislative Assembly', 'VIC Legislative Council']]
```

```python
vic_assembly_vote = {
    'year': np.array([1999, 2002, 2006, 2010, 2014, 2018, 2022]),
    'vote': np.array([1.15, 9.73, 10.04, 11.21, 11.48, 10.71, 11.50])
}
vic_assembly_vote_df = pd.DataFrame(vic_assembly_vote)
vic_assembly_vote_df = vic_assembly_vote_df.set_index('year')

vic_council_vote = {
    'year': np.array([1999, 2002, 2006, 2010, 2014, 2018, 2022]),
    'vote': np.array([2.23, 10.87, 10.58, 12.01, 10.75, 9.25, 10.32])
}
vic_council_vote_df = pd.DataFrame(vic_council_vote)
vic_council_vote_df = vic_council_vote_df.set_index('year')
```

```python
fig, ax = plt.subplots(figsize=(10,6))
ax2 = ax.twinx()

bottom = np.zeros(len(vic_df))
ax.bar(
    vic_df.index,
    vic_df['VIC Legislative Assembly'],
    1,
    label='Legislative Assembly',
    color='tab:green',
    bottom=bottom
)
ax.bar(
    vic_df.index,
    vic_df['VIC Legislative Council'],
    1,
    label='Legislative Council',
    color='tab:red',
    bottom=bottom + vic_df['VIC Legislative Assembly']
)

ax2.plot(
    vic_assembly_vote_df.index,
    vic_assembly_vote_df['vote'],
    color='tab:green',
    marker='o',
    label='Legislative Assembly'
)
ax2.plot(
    vic_council_vote_df.index,
    vic_council_vote_df['vote'],
    color='tab:red',
    marker='o',
    label='Legislative Council'
)

ax.set_ylabel("Number of MPs at year end")
ax2.set_ylabel("Primary vote (%)")
ax.set_xlabel('year')

ax.legend(title='Number of MPs', loc='upper left')
ax2.legend(title='Primary vote (%)', loc='upper right')

ax.set_xlim([1998.5, 2026.5])
ax.set_ylim(0, 14.5)
ax2.set_ylim(0, 14.5)
ax2.grid(False)

plt.title("Victorian Greens electoral history")
plt.savefig('state_mps_vic.png', dpi=300, bbox_inches='tight')
plt.show()
```

The Victorian Greens first contested a state election in 1999. In 2006, the Legislative Council swtiched to a single transferrable vote system.
